```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df =pd.read_csv("homeprices.csv")
df
```

```python
%matplotlib inline
plt.xlabel("Area")
plt.ylabel("Price")
plt.scatter(df.area,df.price,color='red',marker='x')
```

```python
area = df.drop('price',axis='columns')
```

```python
area
```

```python
price=df.price
```

```python
price
```

```python
#create a linear regression object
reg = linear_model.LinearRegression()
reg.fit(area,price)
```

predict the price for various areas using above model

```python
reg.predict([[3300]])
```

```python
reg.predict([[4000]])
```

```python
# slope/gradient/coeficient 
reg.coef_
```

Example2: Using above model , read areas.csv file and predict the prices

```python
area_df = pd.read_csv("areas.csv")
```

```python
area_df
```

```python
price_pred = reg.predict(area_df)
```

```python
price_pred
```

```python
area_df['prices']=price_pred
area_df
```

```python
area_df.to_csv("prediction.csv",index=False)
```
